```python
students=[]
```

```python
def add_student():
    n=int(input("How Many Student do you want to add::"))
    for i in range(n):
        print("\n Enter Student:",i+1, "Details")
        student_id=input("Enter student ID:")
        name=input("Enter student Name:")
        age=int(input("Enter student Age:"))
        python_marks=float(input("Enter Python Marks:"))
        sql_marks=float(input("Enter SQL Marks:"))
        ml_marks=float(input("Enter Machine Learning Marks:"))
        student={
            "ID":student_id,
            "Name":name,
            "Age":age,
            "Python":python_marks,
            "SQL":sql_marks,
            "ML":ml_marks
        }

        students.append(student)
        print("Student Added successfully..!")
    
```

```python
def display_students():
    if len(students) == 0:
        print("No Student recored found.")
    else:
        print("---------------Student Records-------------------")

    for student in students:
        print("ID    :",student['ID'])
        print("Name  :",student['Name'])
        print("Age   :",student['Age'])
        print("Python:",student['Python'])
        print("SQL   :",student['SQL'])
        print("ML    :",student['ML'])
        print("=====================================")
```

##### search student

```python
def search_student():
    student_id=input("Enter Student ID to search::")
    for student in students:
        if student["ID"]==student_id:
            print("=======STUDENT FOUND========")
            print("ID    :",student['ID'])
            print("Name  :",student['Name'])
            print("Age   :",student['Age'])
            print("Python:",student['Python'])
            print("SQL   :",student['SQL'])
            print("ML    :",student['ML'])
            return

    print("STUDENT NOT FOUND...............!")
```

##### Update student

```python
def update_student():
    student_id=input("Enter Student ID to Update::")
    for student in students:
        if student["ID"]==student_id:
            print("=======STUDENT FOUND========")
            student["Name"]=input("Enter New Name:")
            student["Age"]=input("Enter New Age:")
            student["Python"]=input("Enter New Python Marks:")
            student["Name"]=input("Enter New SQL Marks:")
            student["Name"]=input("Enter New Machine Learning Marks:")
            print("STUDENT UPDATED SUCCESSFULLY!")
            return
    print("STUDENT NOT FOUND..........!")
            
```

##### Delete Student

```python
def delete_student():
    student_id=input("Enter Student ID to Delete::")
    for student in students:
         if student["ID"]==student_id:
             students.remove(student)
             print("STUDENT DELETED SUCCESSFULLY..!")
             return
             print("STUDENT NOT FOUND.....!")

    
```

##### Calculate Result

```python
def calculate_result():
    student_id=input("Enter Student ID:")
    for student in students:
        if student["ID"]==student_id:
            total=student["Python"]+student["SQL"]+student["ML"]
            percentage=total/3

            if percentage >=90:
                grade = "A+"
            elif percentage >=80:
                grade = "A"
            elif percentage >=70:
                grade = "B"
            elif percentage >=60:
                grade = "C"
            elif percentage >=50:
                grade = "D"
            elif percentage >=40:
                grade = "D"
            else:
                print("FAILED.........!")
            print("-----STUDENT RESULT--------")
            print("Name:",student['Name'])
            print("Percentage:",percentage)
            print("Grade:",grade)
            return

            print("STUDENT NOT FOUND.")


```

```python
while True:
    print("\n ------------STUDENT MANAGEMENT-------------")
    print("1 .Add Student")
    print("2. Display All Student")
    print("3. Search Student")
    print("4. Update Student")
    print("5. Delete Student")
    print("6. Calculate Result")
    print("7. Exit")
    choice = input("Enter Your Choice:")

    if choice == "1":
        add_student()
    elif choice == "2":
        display_students()
    elif choice == "3":
        search_student()
    elif choice == "4":
        update_student()
    elif choice == "5":
        delete_student()
    elif choice == "6":
        calculate_result()
    elif choice == "7":
        print("THANK YOU FOR USING STUDENT MANAGEMENT SYSTEM!")
    else:
        print("Exit")
    
```
