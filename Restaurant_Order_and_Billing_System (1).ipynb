# Multiple food item and quantity

```python
def order_menu():
    orders=[]

    while True:
        print("\n ========ORDER MENU=========")
        print("1. Add Item")
        print("2. View Order")
        print("3. Finish Order")
        choice=input("Enter Your Choice::")
        if choice=="1":
            item=input("Enter food item::")
            if item in menu:
                quantity=int(input("Enter quantity::"))
                amount=menu[item]*quantity 
                order={
                "item":item,
                "quantity":quantity,
                "amount":amount            
                }

                orders.append(order)
                print("ITEM ADDED SUCCESSFULLY...")
            else:
                print("ITEM NOT FOUNT......!")
        elif choice == "2":
            print("\n--------YOUR ORDER---------------")
            if len(orders)==0:
                print("NO ITEMS ADDED..")
            else:
                for order in orders:
                    print(order['item'],"x",order["quantity"],"=rs",order["amount"])
        elif choice =="3":
            print("Order completed!")
            return orders
        else:
            print("invalid choice.")
                
                
    
```

```python
orders=order_menu()
```

```python
print(orders)
def calculate_total(orders):
    total=0
    for order in orders:
        total =total+order["amount"]
    return total
```

```python
total=calculate_total(orders)
print("Total Bill Rs :",total)
```

# Discount Calulation

```python
def calculate_discount(total):
    discount=total * 0.10
    final_amount = total - discount

    return discount,final_amount
```

```python
discount,final_amount=calculate_discount(total)

print("Total Bill-Rs   :",total)
print("Discount-Rs     :",discount)
print("Final Amount-Rs :",final_amount)
```

# Generate Bill

```python
def generate_bill(customer_name,mobile_number,orders,total,discount,final_amount):
    print("\n"+"=" * 40)
    print("RESTAURANT ORDER AND BILL")
    print("="*40)
    print("customer Name:",customer_name)
    print("Mobile Number:",mobile_number)
    print("\n -------------ORDER DETAILS---------------")
    for order in orders:
        print(
            order['item'],
            "x",order['quantity'],
            '=rs',order['amount']
        )

        print("\n -----------------------------------------")
        print("Total Bill      :rs",total)
        print("Discount        :rs",discount)
        print("Final Amount    :rs",final_amount)
        print("="*40)
        print("Thank You ! Visit Again")

        print("="*40)
```

```python
generate_bill(customer_name,mobile_number,orders,total,discount,final_amount)
```
